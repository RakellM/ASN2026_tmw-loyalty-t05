# Criação da Feature Store (tables)

```python
## Criação da Feature Store
import pandas as pd
dates = [i.strftime("%Y-%m-%d") for i in pd.date_range('2025-01-01', '2026-06-01', freq='MS')]

feature_names = ['fs_loyalty', 'fs_education']

for feature_name in feature_names:
    with open(f"{feature_name}.sql") as file_open:
        query = file_open.read()
        print(feature_name)

    for i in dates:
        print(i)
        df = spark.sql(query.format(date=i))

        (df.write
        .format("delta")
        .mode("overwrite")
        .option("replaceWhere", f"dtRef = '{i}'")
        .saveAsTable(f"workspace.analytics_rm.{feature_name}"))
```

```python
%sql
-- Check LOYALTY

SELECT 
    dtRef,
    COUNT(*)
FROM workspace.analytics_rm.fs_loyalty
GROUP BY ALL ;

```
